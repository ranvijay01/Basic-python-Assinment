```python
# Reverse the string using while loop
str = "Hello world"
index = len(str) - 1
reversed_str = ""
while index >= 0:
  reversed_str = reversed_str + str[index]
  index = index - 1
print(reversed_str)
```

```python
# factorial of n number given by user
n = int(input("Enter a number "))
fact = 1
while n >= 1:
  fact = fact * n
  n = n-1
print(fact)
```
